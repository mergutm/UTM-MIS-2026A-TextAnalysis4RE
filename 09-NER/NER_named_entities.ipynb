#  NER (Named Entity Recognition)

Reconocimiento de Entidades Nombradas, es una tarea del área de Procesamiento de Lenguaje Natural (NLP), dentro de Machine Learning, cuyo objetivo consiste en identificar y clasificar automáticamente entidades relevantes dentro de un texto.

Considerar el siguiente texto:
> El profesor **Juan Pérez** imparte la asignatura de **Inteligencia Artificial** en la **Universidad Tecnológica de la Mixteca**, ubicada en **Oaxaca**, desde **septiembre de 2024**.

Se pueden reconocer las siguientes entidades


| Entidad                               | Categoría |
|---------------------------------------|-----------|
| Juan Pérez                            | PERSON    |
| Inteligencia Artificial               | MISC      |
| Universidad Tecnológica de la Mixteca | ORG       |
| Oaxaca                                | LOC       |
| septiembre de 2024                    | DATE      |



## Las categorías más comunes son:

| Etiqueta | Significado                        | Ejemplo          |
|----------|------------------------------------|------------------|
| PERSON   | Persona                            | Juan Pérez       |
| ORG      | Organización                       | Google           |
| LOC      | Localización                       | Oaxaca           |

| TIME     | Hora                               | 10:30            |

| PERCENT  | Porcentaje                         | 25%              |
| PRODUCT  | Producto                           | iPhone           |
| EVENT    | Evento                             | Juegos Olímpicos |

## Implementación de NER utilizando Machine Learning

Existen diferentes aproximaciones:

### A. Métodos tradicionales de ML

Utilizar algoritmos como:
- Conditional Random Fields (CRF).
- Hidden Markov Models (HMM).
- Support Vector Machines (SVM).

Estos métodos requieren diseñar características (features) lingüísticas, como palabras, prefijos, sufijos, contexto y etiquetas gramaticales.

### B. Deep Learning

Utilizar redes neuronales para aprender representaciones contextuales del lenguaje:
- BiLSTM.
- BiLSTM + CRF.
- CNN + BiLSTM + CRF.


### C. Transformers

Actualmente, una alternativa importante consiste en utilizar modelos preentrenados mediante Transformers:
- BERT.
- BETO (BERT en español).
- RoBERTa.
- XLM-RoBERTa.
- GLiNER.

Estos modelos pueden ajustarse (fine-tuning) con datasets anotados para reconocer entidades específicas.

## Uso de NER

| Aplicación                | Utilidad                                                        |
|---------------------------|-----------------------------------------------------------------|
| Análisis de noticias      | Identificar personas, organizaciones y lugares.                 |
| Análisis de sentimientos  | Asociar opiniones con entidades específicas.                    |
| Sistemas de búsqueda      | Recuperar documentos por personas, instituciones o lugares.     |
| Análisis académico        | Identificar materias, docentes y departamentos.                 |
| Sistemas médicos          | Reconocer enfermedades, medicamentos y tratamientos.            |
| Chatbots                  | Identificar entidades mencionadas en las consultas.             |
| Extracción de información | Transformar documentos no estructurados en datos estructurados. |

```python
import spacy

# Cargar modelo en español
nlp = spacy.load("es_core_news_sm")

texto = """
El profesor Juan Pérez trabaja en la
Universidad Tecnológica de la Mixteca,
ubicada en Oaxaca, México.
"""

# Procesar el texto
doc = nlp(texto)

# Identificar entidades
for entidad in doc.ents:
    print(
        entidad.text, " : ",
        entidad.label_,
        spacy.explain(entidad.label_)
    )
```

# implementación de NER usando HF


**aggregation_strategy="simple"**, busca agrupar los tokens consecutivos que pertenecen a una misma entidad.


```python
from transformers import pipeline

# Identificador del modelo
modelo = "dccuchile/bert-base-spanish-wwm-cased-finetuned-ner"
modelo = "mrm8488/bert-spanish-cased-finetuned-ner"
# Crear pipeline para reconocimiento de entidades
ner = pipeline(
    "token-classification",
    model=modelo,
    aggregation_strategy="first"  # first/max
)

# Texto de entrada
texto = """
El profesor Juan Pérez trabaja en la
Universidad Tecnológica de la Mixteca,
ubicada en Oaxaca, México.

La institución ofrece la carrera de Ingeniería en Computación.
"""

# Ejecutar NER
entidades = ner(texto)

# Mostrar resultados
for entidad in entidades:
    print(
        f"Entidad: {entidad['word']:<30}"
        f" | Tipo: {entidad['entity_group']}"
        f" | Confianza: {entidad['score']:.4f}"
    )
```

```python
import re
from gliner import GLiNER

# Cargar modelo multilingüe (ligero y rápido en CPU/GPU)
modelo = GLiNER.from_pretrained("urchade/gliner_multi-v2.1")
```


### Reglas al redactar etiquetas

GLiNER entiende el significado en lenguaje natural de las palabras  en la lista, la forma en que se escriben afecta directamente la precisión:
* Usar sustantivos en singular: usar "persona", "universidad" o "materia", en lugar de plurales ("personas", "materias"). Los modelos fueron entrenados principalmente con sustantivos singulares.
* Desambiguar conceptos polisémicos: Usar solo "carrera", el modelo puede confundirse con una carrera de autos o trayectoria profesional. Usar "carrera universitaria" o "licenciatura" amarra el significado al contexto académico.
* Evitar etiquetas que se traslapen: usar "lugar" y al mismo tiempo "ciudad o estado" y "pais", competirán entre sí y bajarán la confianza (score). Es mejor mantenerlas mutuamente excluyentes: "ciudad o estado" vs "pais".
* Como todas las etiquetas se inyectan junto al texto en la ventana de contexto del modelo, poner demasiadas (más de 15–20) consume tokens de entrada y vuelve más lenta la inferencia.

```python
texto = """
El profesor Dr. Juan Pérez (ID: 2024-UTM) trabaja en la
Universidad Tecnológica de la Mixteca, ubicada en Oaxaca, México,
desde el 15 de agosto de 2018.

La institución ofrece la carrera de Ingeniería en Computación
y la materia de Inteligencia Artificial.
"""

# Limpiar saltos de línea innecesarios
texto_limpio = re.sub(r'\s+', ' ', texto).strip()

print("texto_limpio: ", texto_limpio)

# Definir exactamente las etiquetas que necesitas
etiquetas = [
    "persona",
    "titulo academico",
    "universidad",
    "ciudad o estado",
    "pais",
    "carrera universitaria",
    "materia",
    "fecha",
    "codigo o identificador"
]

# Predecir entidades (threshold controla la exigencia)
entidades = modelo.predict_entities(texto_limpio, etiquetas, threshold=0.45)

for entidad in entidades:
    print(
        f"Entidad: {entidad['text']:<38}"
        f" | Tipo: {entidad['label']:<22}"
        f" | Confianza: {entidad['score']:.4f}"
    )
```
