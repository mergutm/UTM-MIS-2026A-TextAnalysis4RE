## Probabilidad condicional

La probabilidad condicional representa la posibilidad de que ocurra un evento considerando que otro evento ya ocurrió.

Se expresa como:

$
P(A|B)=\frac{P(A\cap B)}{P(B)}
$

Donde:
- $P(A|B)$: probabilidad de A dado que ocurrió B.
- $P(A\cap B)$: probabilidad de que ocurran A y B.
- $P(B)$: probabilidad de B.


### Ejemplo
Supóngase que se tiene un conjunto de noticias y se observa que la palabra elecciones aparece frecuentemente en noticias de política.

Se desea calcular:

$
P(Política|elecciones)
$

Es decir, la **probabilidad de que una noticia pertenezca a la categoría Política sabiendo que contiene la palabra elecciones.**

## Ejemplo
Supóngase que se cuenta con los siguientes datos meteorológicos:
- En el 30% de los días llueve.
- En el 70% de los días no llueve.
- Cuando llueve, el cielo está nublado el 80% de las veces.
- Cuando no llueve, el cielo está nublado el 20% de las veces.

| Evento |	Probabilidad |
| ---  | ---  |
| Llueve  	| 30% |
| No  llueve	|70% |
| Nublado  cuando llueve	| 80% |
| Nublado  cuando no llueve	| 20% |

Si observamos que el cielo está nublado, **¿cuál es la probabilidad de que esté lloviendo?**

### Identificar los elementos del teorema de Bayes
Definir dos eventos:
- $A$ : Está lloviendo.
- $B$ : El cielo está nublado.

Por tanto:

| Expresión	| Valor |	Interpretación |
| ---	| --- |	--- |
| $P(A)$ 	      | 0.30  |	Probabilidad de lluvia. |
| $P(\neg A)$ 	  | 0.70  |	Probabilidad de que no llueva. |
| $P(B\|A)$ 	  | 0.80  |	Probabilidad de cielo nublado cuando llueve. |
| $P(B\|\neg A)$  | 0.20  |	Probabilidad de cielo nublado cuando no llueve. |

Se desea calcular:

$
P(A|B)
$

Utilizar la ley de probabilidad total:


$
P(B)=P(B|A)P(A)+P(B|\neg A)P(\neg A)
$

Sustituir los valores:

$
P(B)=(0.80)(0.30)+(0.20)(0.70)
$


$
P(B)=0.24+0.14
$


$
\boxed{P(B)=0.38}
$

Ahora sustituir en el teorema de Bayes:

$
P(A|B)=\frac{(0.80)(0.30)}{0.38}
$


$
P(A|B)=\frac{0.24}{0.38} = 0.6316
$

## Suavizado de Laplace (Laplace Smoothing) en Naive Bayes
El suavizado de Laplace es una técnica estadística que permite evitar que una característica tenga probabilidad cero cuando no aparece en los datos de entrenamiento de una categoría.


#### Supóngase que se tienen las siguientes noticias clasificadas:
| Categoría | 	Texto| 
| --- | --- | 
| Deportes | 	equipo fútbol jugador partido| 
| Deportes | 	jugador anotó gol partido| 
| Política | 	gobierno presidente elecciones| 
| Política | 	gobierno elecciones propuesta| 

Se desea calcular la probabilidad de encontrar la palabra fútbol en cada categoría.

Para Política, la palabra fútbol nunca apareció.


$
P(fútbol|Política) = \frac{\text{Frecuencia de fútbol en Política}}{\text{Total de palabras en Política}}
$

Como fútbol no aparece:

$
P(fútbol|Política)=\frac{0}{6}=0
$ 



Supóngase que llega una noticia:
**El presidente presentó una propuesta sobre fútbol**.

Para calcular la probabilidad de que pertenezca a Política, Naive Bayes multiplica las probabilidades de sus palabras:

$
P(Política|D)\propto
P(Política)\times
P(presidente|Política)\times
P(propuesta|Política)\times
P(fútbol|Política)
$

significa "La probabilidad de que el tema sea Política dado el documento \(D\) es proporcional al producto de la probabilidad general de la Política por las probabilidades de que aparezcan sus palabras individuales." 

Como una de las probabilidades es cero, se tiene:

$ P(Política|D)=0 $

El resultado será cero independientemente de las otras probabilidades.

Esto puede provocar que una noticia no sea clasificada correctamente simplemente porque contiene una palabra que no apareció en los documentos de entrenamiento de una categoría.

La solución consiste en asignar una pequeña cantidad de frecuencia a todas las palabras del vocabulario, incluso cuando no hayan aparecido en una categoría.

En un cálculo de probabilidad tradicional, la fórmula es:

$P(x_{i}|{}y)=\frac{\text{Conteo}(x_{i}\text{\ en\ la\ clase\ }y)}{\text{Total\ de\ elementos\ en\ la\ clase\ }y}$


Con el Suavizado de Laplace, la fórmula se modifica sumando un valor (generalmente $1$) al numerador y el tamaño del vocabulario al denominador:

$P(x_{i}|{}y)=\frac{\text{Conteo}(x_{i}\text{\ en\ la\ clase\ }y)+\alpha }{\text{Total\ de\ elementos\ en\ la\ clase\ }y+\alpha \cdot |{}V|{}}$

* $\alpha $ (Alpha): El parámetro de suavizado. En el suavizado de Laplace clásico, $\alpha = 1$. (Si $\alpha < 1$, se le conoce como Suavizado de Lidstone).
* $\vert{}V\vert{}$: El tamaño del vocabulario (el número total de características únicas en todo el conjunto de datos).


Los componentes de los datos de entrenamiento:
1. El Vocabulario (|V|): Es la lista de palabras únicas en todo el entrenamiento.
	* Palabras: equipo, fútbol, jugador, gol, partido, gobierno, elecciones, presidente, propuesta, actor, película, premio.
	* Tamaño del Vocabulario (|V|): 12 palabras únicas.
2. Conteo de palabras por categoría:
	* Deportes: 6 palabras totales (equipo, fútbol, jugador, gol, partido).
	* Política: 5 palabras totales (gobierno, elecciones, presidente, propuesta).
	* Entretenimiento: 5 palabras totales (actor, película, premio).

Aplicando el Suavizado de Laplace (α = 1) para algunos casos:

* Para la categoría Política:
	* Denominador común: $6 \text{ (palabras en Política)} + 12 \text{ (vocabulario)} = 18$
	* $P(\text{presidente} \vert{} \text{Política}) = \frac{1 + 1}{18} = \frac{2}{18} = 0.1111$
	* $P(\text{propuesta} \vert{} \text{Política}) = \frac{1 + 1}{18} = \frac{2}{18} = 0.1111$
	* $P(\text{fútbol} \vert{} \text{Política}) = \frac{0 + 1}{18} = \frac{1}{18} = \mathbf{0.05555}$ (No es cero)
* Para la categoría Deportes:
	* Denominador común: $5 \text{ (palabras en Deportes)} + 12 \text{ (vocabulario)} = 17$
	* $P(\text{presidente} \vert{} \text{Deportes}) = \frac{0 + 1}{17} = \frac{1}{17} \approx 0.0588$
	* $P(\text{propuesta} \vert{} \text{Deportes}) = \frac{0 + 1}{17} = \frac{1}{17} \approx 0.0588$
	* $P(\text{fútbol} \vert{} \text{Deportes}) = \frac{1 + 1}{17} = \frac{2}{17} \approx 0.1176$

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
```

##  Datos de entrenamiento

```python
# Datos (X)
textos_entrenamiento = [
    "equipo fútbol jugador partido",
    "jugador anotó gol partido",
    "gobierno presidente elecciones",
    "gobierno elecciones propuesta"
]
# Etiquetas (y)
categorias_entrenamiento = ["Deportes", "Deportes", "Política", "Política"]
```

```python
# 2. Convertir texto a matriz de conteo de palabras (Vocabulario)
vectorizador = CountVectorizer(token_pattern=r'(?u)\b\w+\b')
X_train = vectorizador.fit_transform(textos_entrenamiento)

# 3. Entrenar el modelo Naive Bayes con Suavizado de Laplace (alpha=1.0)
modelo = MultinomialNB(alpha=1.0)
modelo.fit(X_train, categorias_entrenamiento)

# 4. Nueva noticia a clasificar
nueva_noticia = ["El presidente presentó una propuesta sobre fútbol"]
X_test = vectorizador.transform(nueva_noticia)
```

```python
X_test.toarray()
```

```python
vectorizador.get_feature_names_out()
```

```python
# 5. Predicción
prediccion = modelo.predict(X_test)
probabilidades = modelo.predict_proba(X_test)
```

```python
print("El vocabulario detectado es:", vectorizador.get_feature_names_out())
```

```python
print(f"Noticia clasificada como: {prediccion[0]}")
print(f"Probabilidades: {dict(zip(modelo.classes_, probabilidades[0]))}")
```

## PROCESAMIENTO DE TEXTO Y CLASIFICACIÓN: SPAM


```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
import pandas as pd

# Frases para entrenar
corpus = [
    "me encanta comprar en esta tienda gratis",
    "oferta especial de ganar dinero gratis hoy",
    "reunión de trabajo urgente para mañana",
    "por favor enviar el reporte del proyecto hoy",
    "ganar dinero rápido con esta oferta especial"
]
# Etiquetas: 1 para Spam, 0 para No Spam 
y = [1, 1, 0, 0, 1]

print("Procesamiento de Texto con CountVectorizer (Bag of Words)")
# CountVectorizer crea la representación Bag of Words contando frecuencias absolutas
count_vectorizer = CountVectorizer()
X_counts = count_vectorizer.fit_transform(corpus)

# Mostrar la matriz de conteos en un DataFrame para visualizar
df_counts = pd.DataFrame(X_counts.toarray(), columns=count_vectorizer.get_feature_names_out())
print("Matriz de conteo (BoW):")
print(df_counts.to_string())

```

```python
print("\n Procesamiento de Texto con TF-IDF ")
# TfidfVectorizer calcula la importancia relativa de cada palabra
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(corpus)
```

```python
df_tfidf = pd.DataFrame(X_tfidf.toarray(), columns=tfidf_vectorizer.get_feature_names_out())
print("Matriz TF-IDF (Valores normalizados):")
print(df_tfidf.round(3).to_string())
```

## ENTRENAMIENTO Y CLASIFICACIÓN CON MULTINOMIAL NB

```python
print("Clasificación usando MultinomialNB ")

# modelo para los conteos (BoW)
clf_counts = MultinomialNB()
clf_counts.fit(X_counts, y)

# modelo para TF-IDF
clf_tfidf = MultinomialNB()
clf_tfidf.fit(X_tfidf, y)

# Frases de prueba para clasificar
test_phrases = [
    "obtener dinero gratis ahora",
    "enviar el reporte de la reunión"
]

# Transformar las frases de prueba
X_test_counts = count_vectorizer.transform(test_phrases)
X_test_tfidf = tfidf_vectorizer.transform(test_phrases)

# Predicciones
preds_counts = clf_counts.predict(X_test_counts)
preds_tfidf = clf_tfidf.predict(X_test_tfidf)

# Resultados
for i, phrase in enumerate(test_phrases):
    print(f"\nFrase: \"{phrase}\"")
    print(f"  - Predicción con Bag of Words (CountVectorizer): {'Spam' if preds_counts[i] == 1 else 'No Spam'}")
    print(f"  - Predicción con TF-IDF: {'Spam' if preds_tfidf[i] == 1 else 'No Spam'}")
    
    
```
